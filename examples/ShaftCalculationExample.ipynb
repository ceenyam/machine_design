### dynamic safety factor

the dynamic safety analysis is done using the equations for figure 11-23.  Additionally, as a simplification the construction factor for bending  is also used for torsion.  Since $ \beta_{k\sigma} \gt \beta_{k\tau} $ and $ K_{o\sigma} \lt K_{o\tau}  $, the end result will be on the safe side.  
The calculations uses the amplitude of the occurring stresses. Since a swelling torsion is to be used, $T_{a \ eq} = {T_{nom} \over 2}$

* $ M_{a \ eq} = K_A \cdot M_a = K_A \cdot M_{nom} = 3380 Nm$
* $ \sigma_{ba} = {M_{a \ eq} \over W} = 47.2 {N \over mm^2} $ 
* $ K_{\sigma} = ({{\beta_{k\sigma}\over K_g}+ {1 \over K_{o\sigma}}-1}) \cdot K_v  = 1.89$ 
    * $\beta_{k\sigma} = 1 + c_b(\beta_{k\sigma (test)}-1) = 1+0.4(2.25-1) = 1.5$  using tables 3-9 to 3-12
    * $ K_g = 0.83$ table 3-11c
    * $ K_{o\sigma} = 0.92$ table 3-10 with  $ R_z = 6.3 \mu m $ 
* $ \sigma_{G_W} = {\sigma_{b_W} \over K_{\sigma}} = 152 {N \over mm^2}$
* $ T_{a \ eq} = K_A \cdot T_{a} = 2210 Nm$
* $ \tau_{ta} = {T_{a \ eq}  \over W_p} = 15.4 {N \over mm^2} $ 
* $ K_{\tau} = K_{\sigma} $ 
* $ \tau_{G_W} = {\tau_{t_W} \over K_{\tau}} = 94 {N \over mm^2}$

## exercise 4.3: shaft diameter and safety calculation

![](images/ShaftExample2.png)

<br><br>
The grain elevator in the figure above contains a shaft that requires proper dimensioning. It should be made of E295.<br>
The amount of grain to be transported Q = 50 tonnes/hour and the transport height h = 30 m require a power from the driving engine of P = 7,5 kW at n = 80 rpm. <br>
The efficiency of the gearbox in between the driven shaft and the engine is η = 80 %. <br>
The load factor KA should be taken as 1.2. <br>
Due to construction preferences, dimensions for the disk diameter DS = 800 mm and bearing distance la = 560 mm have already been established. <br>
From previous tests it was concluded that at nominal load the driven shaft is loaded by a radially directed total force (own weight + grain) of: FTot = F1 + F2 ≈ 9,2 kN. <br>
Seeing as the machine should be able to start running with full load (all coops filled) it is reasonable to assume that (since there is no run-in coupling present) the run-in load can be far higher than the
nominal load. In this exercise we can assume this to be by a factor of approximately 2,5. 
<br><br>
To be determined separately:
<ol type="a">
  <li>The diameter d of the driven shaft has to be determined using the global approach. It then has to be checked for strength</li>
  <li>The diameter d1 of the connection stub to the coupling (leading to the gearbox) has to be determined constructively (i.e. standard sizes) and checked.</li>

### overall general comments
Because the distance between teh bearings is known, type 1 from paragraph 11.2.2.-2 is applicable.  Torque is transferred using a coupler, the end of the shaft is therefore only subjected to torque and to bending. Between bearing B and the section A-B, there is a combined loading of both bending and torsion (shear forces are neglected).  Between A-B and bearing A only bending applies. <br>
The calculation of the design diameter is to be done using equation 11-16 from figure 11-21, section A-B is the critical section. <br>
On section A-B, both static and dynamic safety factors have to be determined. <br>
On section C-D only a static safety factor is required: torque is assumed to be static. 


```python
# import statements
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

import MechDesign.Helpers as HM

from MechDesign.Units.Units import m_, mm_, kg_, s_, N_, rpm_, W_, deg_, mu_m_
import MechDesign.Units.UnitMethods as UM

import MechDesign.RnM as RnM
```

```python
# section with constants from the assignment itself
P_motor = 7500*W_   
n = 80 * rpm_
eta = 0.8 #80% efficiency
Q = 50
h = 30*m_
K_A = 1.2
D_s = 800*mm_
l_a = 560*mm_
l_t = 100*mm_  # not specified in the text, a value is assumed for plotting
F_tot = 9200*N_

```

### stting up the basics
the assignment is translated in V, M, T equations.
<br>
**NOTE: we do not yet set $K_A$ but keep it as a symbol.  The value of $K_A$ is depending on what we are calculating: initial estimations of the diameter or Safety Factors.**

```python
s2=RnM.Shaft()
HM.EqPrint('t',s2.E11_10A_TorsionalMoments())

# initialise a new shaft
S = RnM.Shaft()

S.P = P_motor * eta
S.n = n

#!!! note: we do not yet set K_A, but keep it as a symbol.  K_A = 1.2 for the initial estimation of dprime and K_A will be 2.5 when calculating the safety factors. 

F_tot = F_tot * S.K_A
F_1=F_tot/2
F_2=F_tot/2

S.T_nom = S.P / S.n   # not using equations 11.10 because they assume a given set of units

T_eq = UM.All_to_SI( S.E11_11A_TorsionalMoments())  # using E11-11A because it does not rely on a given unit. 
T_eq = UM.kgm_s2_to_N(T_eq)


x = sp.Symbol('x')
V = sp.Piecewise((F_1,x<=l_a/2),(F_1-F_tot,x<=l_a),(F_1-F_tot+F_2,x<=l_a+l_t))
temp = sp.integrate(V,x)    
M = sp.Piecewise(*temp.args[:-1])    #removing NaN part that is generated by the sp.integrate section
S.T = sp.Piecewise((0,x<=l_a/2),(T_eq,x<=l_a+l_t))
HM.EqPrint('V',V)
HM.EqPrint('M',M)
HM.EqPrint('T',S.T)
t=0
```

Note the use of units in the equations. <br>
--- ---
The next step is to calculate $M_V$, using equation 11-7. <br>
Because of the combined load we will set 
${\sigma \over {\\phi \cdot \tau}} = 0.7 $ 

```python
S.M_b = M
S.phi = 1
S.M_v = S.E11_7A_MomentOfComparison()
S.M_v = HM.substitute(S.M_v,S)
t=HM.EqPrint('M_v',S.M_v)
```

### initial estimation of the diameter
using the K_A of 1.2 given, we can calculate the minimum required diameter. <br>

* $K_t = 1 ($ since no diameter is known we first estimate $K_t = 1$)
* $\sigma_{bWN} = 245 {N \over mm^2}$ (E295 in table 1-1)

```python
K_t = 1
sigma_bWN = 245 * N_/mm_**2
K_A = 1.2
S.sigma_bToel = K_t * sigma_bWN
S.sigma_max = S.sigma_bToel
S.tau_max  = S.sigma_max  *  1.2/(3**(1.3))

# creating custom functions to be plotted, each time verify if the units are what you want them to be
# dprime = 3.4*sp.cbrt(S.M_v/S.sigma_bToel)
V1 = HM.substitute(V,S,K_A=K_A)
HM.EqPrint('V1',V1)
M1_b = HM.substitute(S.M_b,S,K_A=K_A)
HM.EqPrint('M1_b',M1_b)
T1 = UM.m_to_mm(HM.substitute(S.T,S,K_A=K_A))
HM.EqPrint('T1',T1)
M_v = UM.m_to_mm(HM.substitute(S.M_v,S,K_A=K_A))
HM.EqPrint('M_v',M_v)
dprime = 3.4*sp.cbrt(S.M_v/S.sigma_bToel)
dprime = HM.substitute(dprime,S,K_A=K_A)
HM.EqPrint('dprime',M_v)


# generating the x values for graphing
l_shaft = float(UM.RemoveUnits(l_a+l_t))
x_val = np.linspace(0,l_shaft,500)


y_V = HM.evaluate(V1,{'x':x_val})
y_M = HM.evaluate(M1_b,{'x':x_val})/1000
y_T = HM.evaluate(T1,{'x':x_val})/1000
y_M_v = HM.evaluate(M_v,{'x':x_val})/1000
y_dprime = HM.evaluate(dprime,{'x':x_val})

# V,M,T graph
fig1,ax1 = plt.subplots(4,1,figsize=(15, 8)) 
line1 = ax1[0].plot(x_val,y_V,label='V (N)')
line2 = ax1[1].plot(x_val,y_T,label='T (Nm)')
line3 = ax1[2].plot(x_val,y_M,label='M_b (Nm)',linestyle=':')
line4 = ax1[2].plot(x_val,y_M_v,label='M_v (Nm)')
line5 = ax1[3].plot(x_val,y_dprime,label=f"$d'$ (mm)")
for a in ax1: 
    a.legend()
    a.grid() 
fig1.suptitle('initial estimation of diameter')
ax1[3].set_xlabel('shaft length (mm)')
t=0
```

As can be seen from the graph, a minimum 65mm is required

### design a shaft
let us assume a shaft of 65mm, stepped down to 50mm at the location of the bearing, reducing further to 48 to go to the coupling.   (this is a different solution compared to the original solution, the stepped down ratio $ {D  \over d} \lt 1.4 $ (section 11.2.1-1)
<br>
let us assume a length of 25mm for the bearings
Assuming also a key connection on the 65mm shaft.  When using a 65mm shaft key depth $t_1 = 7 mm$ (table 12-2)

```python
# Create a shaft 
l_bearing = 25*mm_
d = sp.Piecewise((55*mm_,x<=l_bearing),(65*mm_,x<=l_a),(55*mm_,x<=l_a+l_bearing),(48*mm_,x<=l_a+l_t))
t_1 = sp.Piecewise((0,x<=l_bearing),(7*mm_,x<=l_a),(0,x<=l_a+l_t))  #used to also plot the diameter corrected for the key

fig2,ax2 = plt.subplots(1,1,figsize=(15, 8)) 
y_d = HM.evaluate(d,{'x':x_val})
y_dt1 = HM.evaluate(d-t_1,{'x':x_val})
ax2.plot(x_val,y_dt1,linestyle=':',c='b')
ax2.plot(x_val,y_d,label='chosen diameter (mm)',c='b')
ax2.plot(x_val,y_dprime,label= f"$d'$ (mm)",c='r')
ax2.legend()
ax2.grid()
ax2.set_xlabel('length of shaft (m)')
ax2.set_ylabel('diameter (mm)')
t=0
```

### determine the Static Safety factor, $S_V$
In order to evaluate the static safety factor a $M_{max}$ and $ T_{max}$ are needed.  No specific values are given and the starting Torque of motor can be a multiple of the nominal torque.   On top, it can be expected that starting conditions of a grain elevator could be harsh do to the fully loaded starting. 
Therefore, we will assume a K_A of 2.5. 

using the initial estimation of 65mm, we can use table 12-2 to determine the key height $t_1 = 7 mm$

in a next step we have to determine $M_v$ and $\sigma_{bD}$ 
* $K_t = 0.92 $ (tensile strength and 65mm diameter in tabel 3-11)
* $\sigma_{bWN} = 245 {N \over mm^2} $ (E295 as a material in table 1-1)
* ${\sigma \over {\\phi \cdot \tau}} = 0.7 $

```python
K_A = 2.5  # new K_A value

K_t = 0.92  # table 3-11, yield strength and 65mm diameter
R_pO2N = 295 * N_/mm_**2
sigma_bv = 1.2*R_pO2N*K_t
tau_tv = sigma_bv/sp.sqrt(3).evalf()
HM.EqPrint('sigma_bv',sigma_bv)
HM.EqPrint('tau_tv',tau_tv)


W_b = sp.pi/32*d**3  # a better alternative would be to set W_b = 0.012*(d+(d-t_1))**3, in accordance with table 11-3
#W_b = 0.012*(d+(d-t_1))**3
HM.EqPrint('W_b',W_b)
W_t = sp.pi/16*d**3  # a better alternative would be to set W_t = 0.2*(d-t_1)**3, in accordance with table 11-3
#W_t = 0.2*(d-t_1)**3
HM.EqPrint('W_t',W_t)

# recalculating with new K_A values
V2 = HM.substitute(V,S,K_A=K_A)
HM.EqPrint('V2',V2)
M2_b = HM.substitute(S.M_b,S,K_A=K_A)
HM.EqPrint('M2_b',M2_b)
T2 = UM.m_to_mm(HM.substitute(S.T,S,K_A=K_A))
HM.EqPrint('T2',T2)

sigma = M2_b/W_b
HM.EqPrint('sigma',sigma)
tau = T2/W_t
HM.EqPrint('tau',tau)


S_V = 1 / sp.sqrt((sigma/sigma_bv)**2+(tau/tau_tv)**2)
HM.EqPrint('S_V',S_V)


x_val2 = x_val[1::]  # since there is no bending moment in the very first point of the shaft, using the original x_val would result in a warning. 
y_S_V = HM.evaluate(S_V,{'x':x_val2})

# creating a plot
fig3,ax3 = plt.subplots(1,1,figsize=(15, 8)) 
y_d = HM.evaluate(d,{'x':x_val2})
y_dt1 = HM.evaluate((d-t_1),{'x':x_val2})
ax3.plot(x_val2,y_dt1,linestyle=':',c='b')
ax3.plot(x_val2,y_d,label='chosen diameter (mm)',c='b')
ax3.plot(x_val,y_dprime,label='first estimation (mm)',c='m')
ax3.legend()
ax3.grid()
ax3.set_xlabel('length of shaft (m)')
ax3.set_ylabel('diameter (mm)')
ax3.set_ylim(0,70)
ax3b = ax3.twinx()
ax3b.plot(x_val2,y_S_V,linestyle='--',label=r'$S_V$')
ax3b.set_ylim(0,7)
ax3b.set_ylabel(r'$S_V$')
ax3b.axhline(1.5,c='r',label=r'$S_{Vmin} = 1.5$',linestyle='-.')
ax3b.legend()
t=0
```

 From the graph above, it is clear that even at it's most vulnerable section the static Safety Factor $S_V \gt 1.5$ proves the design is sound, especially taken into account the $K_A = 2.5$ set in the beginning of this calculation.

 ### dynamic safety factor, $S_D$
 We will calculate the dynamic safety factor, including the effect of the key.  Since we will incorporate the key using stress concentration factors, we do not need to take it into account for $W_b$
 
 * $K_A = 1.2$ (per assignment)
 * $K_t = 1$ (dynamic loading -> line 1 in table 3-11a)
 * $R_{mN} = 470 {N \over mm^2}$ (E295 in tabel 1-1)
 * $\beta_{kb} = 1.7$ (milled keyway in shaft, table 3-8)
 * $R_z = 12.5 \mu m$ (comment next to table 3-10)
 * $K_{O \sigma} = 0.91 $ ($R_{mN} = 470 {N \over mm^2}$ and $R_z = 12.5 \mu m$ in table 3-10a)
 * $K_{g \sigma} = 0.85 $ (bending and diameter 65 in table 3-11c)
 * $K_V = 1 $ (no surface strengthening defined, table 3-12 )
 * $S_z = 1.2$ (table 3-14c)
 * $S_Dmin = 1.5$ (table 3-14a)
 

```python
K_A = 1.2
K_t = 1
R_mN = 470 *N_ / mm_**2
beta_kb = 1.75
R_z = 12.5 *mu_m_
K_O_sigma = 0.91 
K_g_sigma = 0.85 
K_V = 1

S_z = 1.2
S_Dmin = 1.5

R_m = K_t * R_mN
K_sigma = (beta_kb/K_g_sigma + 1/K_O_sigma -1)*1/K_V
HM.EqPrint('K_sigma',K_sigma)

sigma_GW = sigma_bWN * K_t / K_sigma
HM.EqPrint('sigma_GW',sigma_GW)

# recalculate 
# recalculating with new K_A values
V3 = HM.substitute(V,S,K_A=K_A)
HM.EqPrint('V3',V3)
M3_b = HM.substitute(S.M_b,S,K_A=K_A)
HM.EqPrint('M3_b',M3_b)
T3 = UM.m_to_mm(HM.substitute(S.T,S,K_A=K_A))
HM.EqPrint('T3',T3)



sigma_ba = M3_b/W_b
sigma_ba = sigma_ba.evalf()
HM.EqPrint('sigma_ba',sigma_ba)

l_shaft_short = float(UM.RemoveUnits(l_a))
x_val_short = np.linspace(l_shaft_short/500,l_shaft_short-l_shaft_short/500,500)
# generating the x values for graphing
#l_shaft_short = float(UM.RemoveUnits(l_a))
#x_val_short = np.linspace(1,l_shaft_short-1,500)
y_sigma_ba = HM.evaluate(sigma_ba,{'x':x_val_short})


S_D = sigma_GW/sigma_ba
HM.EqPrint('S_D',S_D)
display(S_D.free_symbols)
# replace the values equal to infinity with zero, needed for evaluation to work
S_D = S_D.subs({sp.zoo:0})
HM.EqPrint('S_D',S_D)
```

```python
# creating a plot
fig4,ax4 = plt.subplots(1,1,figsize=(15, 8)) 
y_d = HM.evaluate((d),{'x':x_val2})
y_dt1 = HM.evaluate((d-t_1),{'x':x_val2})
ax4.plot(x_val2,y_dt1,linestyle=':',c='b')
ax4.plot(x_val2,y_d,label='chosen diameter (mm)',c='b')
ax4.plot(x_val,y_dprime,label='first estimation (mm)',c='m')
ax4.legend(loc='upper left')
ax4.grid()
ax4.set_xlabel('length of shaft (mm)')
ax4.set_ylabel('diameter (mm)')
ax4.set_ylim(30,70)
ax4b = ax4.twinx()
y_S_D =  HM.evaluate(S_D,{'x':x_val_short})
ax4b.plot(x_val_short,y_S_D,linestyle='--',label=r'$S_D$')
ax4b.set_ylim(0,7)
ax4b.set_ylabel(r'$S_D$')
ax4b.axhline(S_Dmin*S_z,c='r',label=r'$S_{Dmin} = 1.5$ and $S_z = 1.2$',linestyle='-.')
ax4b.legend(loc='upper right')
ax4.set_xlim([0,660])
t=0
```

as can be seen: when including the stress concentration factor for the key, the $S_D$ stays above or drops below the threshold depending of your choice of W_b and W_t
